# Smart Meal Planner AI: Naïve Bayes Meal Category Classification

This notebook trains a Multinomial Naïve Bayes classifier with Laplace smoothing on user dietary preferences, budget, fitness goals, and time of day to predict the optimal meal category (breakfast, lunch, snack, dinner).

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../data/meal_preferences.csv')
print(df.head())

feature_cols = ['dietary_pref', 'cuisine_pref', 'budget_category', 'fitness_goal', 'activity_level', 'time_of_day', 'protein_pref']
target_col = 'target_meal_category'

X = df[feature_cols]
y = df[target_col]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)

encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
X_train_enc = encoder.fit_transform(X_train)
X_test_enc = encoder.transform(X_test)

clf = MultinomialNB(alpha=1.0)
clf.fit(X_train_enc, y_train)

y_pred = clf.predict(X_test_enc)
print(classification_report(y_test, y_pred))
cm = confusion_matrix(y_test, y_pred, labels=clf.classes_)
print("Confusion Matrix:\n", cm)
```
